```python
df_raw['ret'] = df_raw.groupby('symbol')['close_price'].pct_change()
df_raw['ret'] = df_raw['ret'].fillna(0)

df_sectors = df_raw.groupby(['trading_date', 'industry'])['ret'].mean().unstack()

df_sectors = df_sectors.dropna(how='all')
df_sectors.head(5)
```

```python
df_raw['trading_date'] = pd.to_datetime(df_raw['trading_date'])
df_clean = df_raw[df_raw['symbol'].str.len() == 3].copy()

#Tạo cột mốc cuối tháng
df_clean['month'] = df_clean['trading_date'] + pd.offsets.MonthEnd(0)
df_clean = df_clean.sort_values(['symbol', 'trading_date'])

#Lấy bản ghi cuối cùng trong tháng của mỗi mã
df_monthly = (
    df_clean.groupby(['symbol', 'month'])
    .last()
    .reset_index()[['symbol', 'month', 'close_price', 'industry']]
)

# Tính lợi nhuận hàng tháng (Monthly Return)
df_monthly['ret'] = df_monthly.groupby('symbol')['close_price'].pct_change()


# J=6 tháng kèm Skip-1-Month: từ t-6 đến t-1
def calc_momentum(group):
  # shift(1) để bỏ qua tháng t, rolling(6) tính tích lũy 6 tháng trước đó
  group['cum_ret_form'] = (
      (1 + group['ret']).shift(1).rolling(6).apply(np.prod, raw=True) - 1
  )
  return group


df_monthly = df_monthly.groupby('symbol', group_keys=False).apply(calc_momentum)

# Bỏ các dòng chưa đủ 6 tháng
df_signal = df_monthly.dropna(subset=['cum_ret_form']).copy()
df_signal.head(5)
```

```python
df_raw['month'] = df_raw['trading_date'] + pd.offsets.MonthEnd(0)
industry_portfolio = (
    df_raw.groupby(['industry', 'month'])['ret'].mean().unstack(level=0)
)

# Chỉ giữ các ngành có tối thiểu số quan sát hợp lệ (loại nhóm 'Khác' nếu quá hỗn tạp)
if 'Khác' in industry_portfolio.columns:
  industry_portfolio = industry_portfolio.drop(columns=['Khác'])

print(list(industry_portfolio.columns))
```

```python
def calc_autocorr(series, max_lag=20):
  """Tính autocorrelation cho chuỗi tỷ suất sinh lợi từ lag 1 đến max_lag."""
  autocorrs = {}
  clean_series = series.dropna()
  for k in range(1, max_lag + 1):
    # Tính tương quan giữa chuỗi gốc và chuỗi trễ k kỳ
    ac = clean_series.autocorr(lag=k)
    autocorrs[k] = ac
  return pd.Series(autocorrs)


max_k = 20
df_autocorr = pd.DataFrame(index=range(1, max_k + 1))

for ind in industry_portfolio.columns:
  df_autocorr[ind] = calc_autocorr(industry_portfolio[ind], max_lag=max_k)

display(df_autocorr.round(3))
```

```python
summary_lags = []
for ind in df_autocorr.columns:
  series = df_autocorr[ind].dropna()
  if not series.empty:
    # Lag có quán tính đà tăng mạnh nhất (Positive Autocorrelation lớn nhất)
    best_pos_lag = series.idxmax()
    max_pos_val = series.max()

    # Lag có độ tương quan mạnh nhất nói chung (|r| max)
    abs_best_lag = series.abs().idxmax()
    abs_val = series.loc[abs_best_lag]

    summary_lags.append({
        'Ngành': ind,
        'Lag Momentum cao nhất (Max r)': best_pos_lag,
        'Hệ số r max': round(max_pos_val, 3),
        'Lag tác động mạnh nhất (|r| max)': abs_best_lag,
        'Hệ số |r| max': round(abs_val, 3),
    })

df_summary = pd.DataFrame(summary_lags)
print('\n--- KẾT QUẢ ĐỘ LAG TỐI ƯU THEO TỪNG DANH MỤC NGÀNH ---')
display(df_summary)

# Kiểm tra xem giữa các nhóm ngành có lag chung nào không:
common_lags = df_summary['Lag Momentum cao nhất (Max r)'].value_counts()
print('\n--- TẦN SUẤT XUẤT HIỆN CỦA CÁC ĐỘ LAG CHUNG (MAX POSITIVE) ---')
display(common_lags[common_lags > 1])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 8))

for ind in df_autocorr.columns:
  plt.plot(
      df_autocorr.index,
      df_autocorr[ind],
      marker='o',
      linewidth=1.8,
      alpha=0.85,
      label=ind,
  )

# Đường chuẩn y = 0
plt.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.7)

# Vùng ngưỡng ý nghĩa thống kê sơ bộ 95% Bartlett: +- 1.96 / sqrt(N)
n_obs = len(industry_portfolio)
threshold = 1.96 / np.sqrt(n_obs)
plt.axhline(
    threshold,
    color='red',
    linestyle=':',
    alpha=0.6,
    label=f'Ngưỡng ý nghĩa 95% (±{round(threshold, 2)})',
)
plt.axhline(-threshold, color='red', linestyle=':', alpha=0.6)

plt.title(
    'Autocorrelation of Industry Portfolios Across Lags (k = 1 to 20)',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Lag (k - tháng)', fontsize=12)
plt.ylabel('Autocorrelation Coefficient (r)', fontsize=12)
plt.xticks(range(1, max_k + 1))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', frameon=True)
plt.tight_layout()
plt.show()
```
